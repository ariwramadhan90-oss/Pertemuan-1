```python
nama = "Arif"
umur = 21
tinggi = 173
angka = 255

# %s untuk string
print("Nama saya %s" % nama)

# %d untuk integer
print("Umur saya %d tahun" % umur)

# %f untuk float
print("Tinggi saya %f cm" % tinggi)

# %x untuk hexadecimal
print("Angka %d dalam hexadecimal adalah %x" % (angka, angka))

# %o untuk octal
print("Angka %d dalam octal adalah %o" % (angka, angka))
```

```python
total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
# Menghitung Nilai Akhir Mahasiswa

tugas = 80
uts = 70
uas = 90

nilai_akhir = (0.3 * tugas) + (0.3 * uts) + (0.4 * uas)

print("Nilai akhir =", nilai_akhir)
```

```python
# Input melalui keyboard
nilai_mahasiswa = float(input("Masukkan nilai mahasiswa: "))
standar_kelulusan = float(input("Masukkan standar kelulusan: "))
```

```python
# Menentukan status mahasiswa
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
belanja = int(input("Masukkan total belanja: Rp "))

if belanja > 1000000:
    diskon = 0.35 * belanja
else:
    diskon = 0

total_harga = belanja - diskon

print("Total harga setelah diskon = Rp", total_harga)
```
